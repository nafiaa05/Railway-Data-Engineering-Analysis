# Level 2: Data Transformation and Aggregation

## Task 2.1: Data Filtering

### Objective

- Filter the dataset to display trains that operate only on Saturdays.
- Create a new dataframe containing trains that start from a specific source station.

####  Filter Trains Operating on Saturday

```python
saturday_trains = df[df["days"] == "Saturday"]

print("Number of Saturday Trains:", len(saturday_trains))

saturday_trains.head(10)
```

####  Create a New DataFrame for Trains Starting from CST-MUMBAI

```python
source_station_df = df[df["Source_Station_Name"] == "CST-MUMBAI"]

print("Number of trains starting from CST-MUMBAI:", len(source_station_df))

source_station_df.head(10)
```

### Observation

- Successfully filtered trains operating on Saturdays.
- Created a separate dataframe for trains starting from CST-MUMBAI.
- Data filtering helps retrieve specific records for further analysis.

## Task 2.2: Grouping and Aggregation

### Objective

- Count the number of trains originating from each source station.
- Find the average number of trains per day for each source station.

#### Count Number of Trains from Each Source Station

```python
station_count = df.groupby("Source_Station_Name").size().reset_index(name="Number_of_Trains")

station_count.head(10)
```

#### Average Number of Trains Per Day for Each Source Station

```python
average_trains = (
    df.groupby(["Source_Station_Name", "days"])
      .size()
      .groupby(level=0)
      .mean()
      .reset_index(name="Average_Trains_Per_Day")
)

average_trains.head(10)
```

### Observation

- Grouped the dataset by source station.
- Calculated the total number of trains originating from each station.
- Computed the average number of trains operating per day for each source station.

## Task 2.3: Data Enrichment

### Objective

- Add a new column to categorize trains based on their operating days as Weekday or Weekend.

### Add Weekday and Weekend Category

```python
# Categorize operating days

def categorize_day(day):
    if day in ["Saturday", "Sunday"]:
        return "Weekend"
    else:
        return "Weekday"

df["Day_Category"] = df["days"].apply(categorize_day)

df.head(10)
```

#### Verify Day Categories

```python
df["Day_Category"].value_counts()
```

### Observation

- Added a new column named **Day_Category**.
- Categorized Saturday and Sunday as **Weekend**.
- Categorized Monday to Friday as **Weekday**.
- The enriched dataset is ready for advanced analysis.

# Level 3: Advanced Data Analysis

## Task 3.1: Pattern Analysis

### Objective

- Analyze the distribution of train journeys throughout the week.
- Visualize the results using bar plots.
- Identify patterns in train operations based on source and destination stations.

#### Distribution of Train Journeys Throughout the Week

```python
# Count the number of trains operating on each day
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_distribution = df["days"].value_counts().reindex(day_order)

print(day_distribution)
```

#### Visualize Using a Bar Plot

```python
plt.figure(figsize=(8,5))

day_distribution.plot(
    kind="bar",
    color="orange",
    edgecolor="black"
)

plt.title("Distribution of Train Journeys Throughout the Week", fontsize=14, fontweight="bold")
plt.xlabel("Days")
plt.ylabel("Number of Trains")

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Identify Patterns Based on Source Stations

```python
# Top 10 source stations

top_source = df["Source_Station_Name"].value_counts().head(10)

print(top_source)
```

#### Visualize Top 10 Source Stations

```python
plt.figure(figsize=(10,5))

top_source.plot(
    kind="bar",
    color="purple",
    edgecolor="black"
)

plt.title("Top 10 Source Stations", fontsize=14, fontweight="bold")
plt.xlabel("Source Stations", fontsize=12)
plt.ylabel("Number of Trains", fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Identify Patterns Based on Destination Stations

```python
# Top 10 destination stations

top_destination = df["Destination_Station_Name"].value_counts().head(10)

print(top_destination)
```

####  Visualize Top 10 Destination Stations

```python
plt.figure(figsize=(10,5))

top_destination.plot(
    kind="bar",
    color="green",
    edgecolor="black"
)

plt.title("Top 10 Destination Stations", fontsize=14, fontweight="bold")
plt.xlabel("Destination Stations", fontsize=12)
plt.ylabel("Number of Trains", fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

### Observation

- Train operations are distributed across all days of the week.
- CST-MUMBAI is one of the busiest source stations.
- CST-MUMBAI is also one of the busiest destination stations.
- The bar charts clearly highlight the major railway hubs and train operation patterns.

## Task 3.2: Correlation and Insights

### Objective

- Explore the relationship between train frequency and operating days.
- Provide insights and recommendations based on the analysis.

#### Number of Trains Operating on Each Day

```python
train_per_day = df["days"].value_counts().reindex([
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
])

print(train_per_day)
```

#### Visualize the Relationship

```python
plt.figure(figsize=(8,5))

train_per_day.plot(
    kind="bar",
    color="hotpink",
    edgecolor="black"
)

plt.title("Number of Trains by Operating Day", fontsize=14, fontweight="bold")
plt.xlabel("Days")
plt.ylabel("Number of Trains")

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

### Observation

- Train services operate on all seven days of the week.
- Friday has the highest number of train operations.
- Monday has the lowest number of train operations.
- The difference in train frequency across the week is relatively small, indicating a balanced railway schedule.

## Insights and Recommendations

### Insights

- Train operations are evenly distributed throughout the week.
- Friday experiences the highest train frequency, indicating higher travel demand before weekends.
- The railway schedule is balanced, with only slight variations between weekdays.

### Recommendations

- Increase train availability on high-demand days such as Friday if passenger demand increases.
- Continue monitoring train schedules regularly to improve operational efficiency.
- Maintain balanced train operations to ensure reliable passenger service throughout the week.
